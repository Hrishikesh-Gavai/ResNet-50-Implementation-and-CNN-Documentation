```python
import numpy as np                  # for working with numbers and arrays
import matplotlib.pyplot as plt     # for showing images
from PIL import Image               # for saving the image as a PNG

# Load all van drawings
data = np.load("van.npy")

# Rows = number of drawings, columns = 784 pixels per drawing
print("Shape:", data.shape)
```

```python
plt.figure(figsize=(10, 4))

for i in range(10):
    img = data[i].reshape(28, 28)     # turn 784 numbers back into a 28x28 image

    plt.subplot(2, 5, i + 1)          # 2 rows, 5 columns, position i+1
    plt.imshow(img, cmap="gray")      # show it in black and white
    plt.title("van " + str(i))
    plt.axis("off")                   # hide the axis numbers

plt.show()
```

```python
idx = 7                               # the drawing we want to use

img = data[idx].reshape(28, 28)       # 784 numbers -> 28x28 image

Image.fromarray(img).save("van.png")  # save it as a PNG file

print("Shape:", img.shape)            # should be (28, 28)
print("Smallest value:", img.min())   # 0 = black background
print("Largest value:", img.max())    # 255 = white stroke
```

```python
# Show the image bigger
plt.figure(figsize=(3, 3))
plt.imshow(img, cmap="gray")
plt.title("van " + str(idx))
plt.axis("off")
plt.show()

# Print all 28x28 numbers on one screen without wrapping
print("\n")
np.set_printoptions(linewidth=250)
print(img)
```

```python
# Turn 0-255 values into 0-1 values
image = img / 255.0

np.set_printoptions(linewidth=250, precision=2, suppress=True) # show 2 decimal places

print("Input shape:", image.shape)
print("\n")
print(image)
```

```python
padding = 1

# Add a border of zeros around the image
padded = np.pad(image, padding)

print("Padded shape:", padded.shape)     # 28x28 becomes 30x30
print("\n")
print(padded)
```

```python
# 3x3 vertical edge detector:
# left column negative, right column positive
kernel = np.array([[-1, 0, 1],
                   [-1, 0, 1],
                   [-1, 0, 1]])

stride = 1

print("Kernel:")
print(kernel)
print("\nStride:", stride)
```

```python
k = 3                                            # kernel size
out_size = (28 - k + 2 * padding) // stride + 1  # output size formula
print("Output size:", out_size, "x", out_size)

feature_map = np.zeros((out_size, out_size))     # empty matrix to fill

for i in range(out_size):                        # go down row by row
    for j in range(out_size):                    # go across column by column
        r = i * stride                           # top-left row of the window
        c = j * stride                           # top-left column of the window

        patch = padded[r:r+3, c:c+3]             # take a 3x3 piece of the image

        feature_map[i, j] = np.sum(patch * kernel)   # multiply and add up

print("\n")
print(feature_map)
```

```python
i, j = 6, 10                          # pick one position in the output

patch = padded[i:i+3, j:j+3]          # the 3x3 piece the kernel saw there

print("Image patch:")
print(patch)
print("\nKernel:")
print(kernel)
print("\nPatch x Kernel (element by element):")
print(patch * kernel)
print("\nSum =", np.sum(patch * kernel))
print("Value in feature map =", feature_map[i, j])
```

```python
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(image, cmap="gray")
plt.title("Input (28x28)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(feature_map, cmap="gray")
plt.title("Feature map (28x28)")
plt.axis("off")

plt.show()
```

```python
# ReLU: negative values become 0, positive values stay the same
relu_map = np.maximum(0, feature_map)

print("ReLU output shape:", relu_map.shape)     # still 28x28
print("\n")
print(relu_map)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(image, cmap="gray")
plt.title("Input")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(feature_map, cmap="gray")
plt.title("Feature map")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(relu_map, cmap="gray")
plt.title("After ReLU")
plt.axis("off")

plt.show()
```

```python
pool_size = 2                                    # 2x2 window
pool_stride = 2                                  # jump 2 pixels each time
pool_out = (28 - pool_size) // pool_stride + 1   # same formula as conv (no padding)
print("Pooling output size:", pool_out, "x", pool_out)

pooled = np.zeros((pool_out, pool_out))          # empty 14x14 matrix to fill

for i in range(pool_out):                        # go down row by row
    for j in range(pool_out):                    # go across column by column
        r = i * pool_stride                      # top-left row of the window
        c = j * pool_stride                      # top-left column of the window

        block = relu_map[r:r+2, c:c+2]           # take a 2x2 piece

        pooled[i, j] = np.max(block)             # keep only the biggest value

print("\n")
print(pooled)
```

```python
plt.figure(figsize=(8, 4))

plt.subplot(1, 2, 1)
plt.imshow(relu_map, cmap="gray")
plt.title("After ReLU (28x28)")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(pooled, cmap="gray")
plt.title("After Max Pooling (14x14)")
plt.axis("off")

plt.show()
```

```python
# Lay out all 14x14 values in one long list, row by row
flat = pooled.flatten()

print("Before flatten:", pooled.shape)    # (14, 14)
print("\nAfter flatten: ", flat.shape)      # (196,)
print("\n")
print(flat)
```

```python
np.random.seed(0)                         # same random numbers every time you run it

classes = ["van", "car", "bus"]

weights = np.random.randn(196, 3) * 0.1   # 196 inputs x 3 classes = 588 weights
biases = np.zeros(3)                      # one bias per class, set to 0

print("Weights shape:", weights.shape)    # (196, 3)
print("Biases:", biases)
```

```python
scores = np.zeros(3)                      # empty list for the 3 scores

for k in range(3):                        # for each class
    total = 0
    for n in range(196):                  # go through all 196 inputs
        total = total + flat[n] * weights[n, k]   # input x its weight, keep adding
    scores[k] = total + biases[k]         # add the bias at the end

print("Scores:", scores)
```

```python
from scipy.special import softmax

probs = softmax(scores)                   # does exp, sum, and divide in one step

print("Scores:       ", scores)
print("Probabilities:", probs)
print("Sum:", np.sum(probs))              # should be 1.0

best = np.argmax(probs)                   # position of the highest probability

print("\n")
for k in range(3):
    print(classes[k], ":", round(probs[k] * 100, 2), "%")

print("\nPrediction:", classes[best])
```
